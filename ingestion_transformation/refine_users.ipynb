### Overview
This notebook loads user demographic data with explicit data type specifications for memory efficiency and data integrity, then converts the data to Parquet format for optimized storage and access.


```python
import pandas as pd
```

```python
# Load user demographic data with explicit data types 
df = pd.read_csv("users_edit.csv", dtype={
        'UserID': 'int64',       
        'Age': 'int64',       
        'Gender': 'string',  
        'CAP': 'string',
        'Work': 'string'
    })

pd.set_option('display.max_rows', None)
df
```

```python
# validation of the Gender column
df["Gender"].unique()
```

```python
# create parquet
df.to_parquet("Users.parquet")
```
